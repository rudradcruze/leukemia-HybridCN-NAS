```python
import pandas as pd
import umap
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
file_path = '../2_Feature_Extraction/Hybrid_ConvNeXt_NASNetLarge/feature_extract_Hybrid_features.csv'
data = pd.read_csv(file_path)

# Extract features and labels
features = data.iloc[:, :-3]  # All columns except the last three (label, class, filename)
labels = data['class']

# Apply UMAP
reducer = umap.UMAP()
umap_embedding = reducer.fit_transform(features)

# Create a DataFrame for the UMAP result
umap_df = pd.DataFrame(umap_embedding, columns=['UMAP1', 'UMAP2'])
umap_df['class'] = labels

# Plot the UMAP representation
plt.figure(figsize=(12, 8))
plt.rcParams['font.family'] = 'Times New Roman'
sns.scatterplot(
    x='UMAP1', 
    y='UMAP2', 
    hue='class', 
    data=umap_df, 
    palette='tab10',  # Changed to 'tab10' for more distinct colors
    s=100,           # Increased scatter point size
    legend='full'
)
plt.title('UMAP Representation of the Dataset', fontsize=28, pad=20)
plt.xlabel('UMAP1', fontsize=22)
plt.ylabel('UMAP2', fontsize=22)
plt.xticks(fontsize=18)
plt.yticks(fontsize=18)
plt.legend(loc='best', fontsize=18, markerscale=2)
plt.savefig('UMAP_representation_Hybrid_ConvNeXt_NASNetLarge_v1.png', dpi=1000)
plt.savefig('UMAP_representation_Hybrid_ConvNeXt_NASNetLarge_v1.pdf', dpi=1000)
plt.show()
plt.close()
```
